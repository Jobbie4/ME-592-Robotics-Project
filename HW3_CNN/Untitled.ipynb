```python
import os
import shutil
import csv

# Define the base directory where your test folders are located
base_dir = "data"
dest_folder = os.path.join(base_dir, "test_10")
dest_img_folder = os.path.join(dest_folder, "IMG")

# Create destination folders if they don't exist
os.makedirs(dest_img_folder, exist_ok=True)

# This list will hold rows for the combined CSV data.
combined_csv_rows = []
header_saved = False  # flag to ensure header is only added once

# Loop over directories in the base folder that start with 'test_' (except test_10)
for folder in os.listdir(base_dir):
    if folder.startswith("test_") and folder != "test_10":
        folder_path = os.path.join(base_dir, folder)
        if os.path.isdir(folder_path):
            # Process the CSV file
            csv_path = os.path.join(folder_path, "robot_log.csv")
            if os.path.exists(csv_path):
                with open(csv_path, "r", newline="") as csvfile:
                    reader = csv.reader(csvfile)
                    rows = list(reader)
                    if rows:
                        # If header hasn't been saved, add all rows (including header).
                        # Otherwise, skip the header from subsequent CSV files.
                        if not header_saved:
                            combined_csv_rows.extend(rows)
                            header_saved = True
                        else:
                            combined_csv_rows.extend(rows[1:])  # skip header row

            # Process images in the IMG folder
            img_source_folder = os.path.join(folder_path, "IMG")
            if os.path.isdir(img_source_folder):
                for img_name in os.listdir(img_source_folder):
                    # Check if the file is a .jpg image (case insensitive)
                    if img_name.lower().endswith(".jpg"):
                        src_img_path = os.path.join(img_source_folder, img_name)
                        # Now, simply use the original image name without any prefix.
                        dest_img_path = os.path.join(dest_img_folder, img_name)
                        shutil.copy2(src_img_path, dest_img_path)

# Write the combined CSV data into a new file in test_10 folder
combined_csv_path = os.path.join(dest_folder, "robot_log.csv")
with open(combined_csv_path, "w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerows(combined_csv_rows)

print(f"Combined CSV and images have been saved to: {dest_folder}")
```
